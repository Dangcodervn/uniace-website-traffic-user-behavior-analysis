```python
import os
import pandas as pd
import datetime as dt
```

```python
path = "D:/Data Self Learning/Tran Hoang Long _ Data Analyst Course/Data Analyst/Uniace Website Traffic & User Behavior Analysis/Clean data/"
frames = []
for file in os.listdir(path) :
    if file.endswith(".xlsx") :
        filepath = path + file
        dftmp = pd.read_excel(filepath)
        frames.append(dftmp)
df = pd.concat(frames)
```

```python
def parse_date(x):
    if pd.isna(x):
        return pd.NaT
    if isinstance(x, (pd.Timestamp, dt.datetime)):
        ts = pd.Timestamp(x)
        # Excel da tu doc nham chuoi "DD-08-2021" (ngay-thang) theo kieu My
        # (thang-ngay), luon bien thang goc (08) thanh "ngay" va ngay goc (1-12)
        # thanh "thang" -> nhan dien qua dau hieu ngay=8, thang!=8, roi doi lai
        if ts.day == 8 and ts.month != 8:
            try:
                return ts.replace(month=8, day=ts.month)
            except ValueError:
                return pd.NaT
        return ts
    s = str(x).strip()
    # Thu lan luot cac dinh dang da phat hien trong du lieu goc,
    # luon uu tien ngay-thang-nam (quy uoc Viet Nam), khong de pandas tu doan
    for fmt in ('%d-%m-%Y %H:%M:%S', '%d/%m/%Y %H:%M', '%Y-%m-%d %H:%M:%S'):
        try:
            return pd.to_datetime(s, format=fmt)
        except ValueError:
            continue
    return pd.NaT

df['Date'] = df['Date'].apply(parse_date)
df = df[df['Date'].notna()]
```

```python
df.to_csv(r"D:/Data Self Learning/Tran Hoang Long _ Data Analyst Course/Data Analyst/Uniace Website Traffic & User Behavior Analysis/Clean data/finaldata.csv", index=False, encoding='utf-8')
```
