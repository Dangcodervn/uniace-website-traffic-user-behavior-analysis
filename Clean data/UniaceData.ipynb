```python
import os
import pandas as pd
import datetime as dt
```

```python
path = "D:/study_da/Uniace analytic Project/Clean data/"
frames = []
for file in os.listdir(path) :
    if file.endswith(".xlsx") :
        filepath = path + file
        dftmp = pd.read_excel(filepath)
        frames.append(dftmp)
df = pd.concat(frames)
```

```python
df = df[pd.to_datetime(df['Date'], errors='coerce').notna()]
df['Date'] = pd.to_datetime(df['Date'])
```

```python
df['Day'] = df['Date'].dt.date
df['Hour'] = df['Date'].dt.hour
```

```python
df.to_csv(r"D:/study_da/Uniace analytic Project/Clean data/finaldata.csv", index=False, encoding='utf-8')
```
